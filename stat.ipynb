```python
import statistics
import numpy as np

data = [2, 4, 6, 8, 10]

# Mean
print("Mean-1 =", statistics.mean(data))

print("Mean-2 =", np.mean(data))

# Median
print("Median-1 =", statistics.median(data))

print("Median-2 =", np.median(data))

# Mode
print("Mode =", statistics.mode(data))

# Range
print("Range =", max(data) - min(data))

# IQR
print("IQR =", np.percentile(data, 75) - np.percentile(data, 25))
# from scipy.stats import iqr
# print(iqr(data))

# Variance
print("Variance-1 =", statistics.pvariance(data))
print("Variance-2 =", statistics.variance(data))
print("Variance-3 =", np.var(data))

# Standard Deviation
print("Standard Deviation-1 =", statistics.pstdev(data))
print("Standard Deviation-2 =", statistics.stdev(data))
print("Standard Deviation-3 =", np.std(data))

# Deviation, Absolute Deviation, Mean Absolute Deviation
mean = np.mean(data)
print("Deviation-1 =", data - mean) # for numpy do this (2 lines)

print("Absolute Deviation =", np.abs(data - mean))

print("Mean Absolute Deviation-1 =", np.mean(np.abs(data - mean)))

# Mean Absolute Deviation
mean = statistics.mean(data)

deviation = [x - mean for x in data] # for statistics do this (3 lines)
print("Deviation-2 =", deviation)

MAD = statistics.mean([abs(x - mean) for x in data])
print("Mean Absolute Deviation-2 =", MAD)

# Covariance
X = [1, 2, 3, 4]
Y = [40, 50, 60, 70]
print("Covariance_1 =", np.cov(X, Y, ddof=0)[0][1])

covariance = statistics.covariance(X, Y)
print("Covariance_2 =", covariance)

# correlation
study_hours = [3, 2, 5, 7, 1, 3, 3.5, 2.2]
marks = [50, 40, 70, 90, 15, 48, 65, 45]
x = statistics.correlation(study_hours, marks)
print(round(x, 2))

# Skewness
print("Skewness =", np.mean((data - np.mean(data))**3) /
      np.std(data)**3)

# Kurtosis
kurtosis = np.mean((data - np.mean(data))**4) / np.std(data)**4
print("Kurtosis =", kurtosis)

# Excess Kurtosis
print("Excess Kurtosis =", kurtosis - 3)
```
